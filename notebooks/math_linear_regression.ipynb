```python
y = df['sales']
X = df.drop('sales', axis=1)
```

**Классический подход**

```python
X
```

```python
y
```

```python
X.corr()
```

```python
X.shape[0]
```

```python
n = X.shape[0]
X = np.column_stack((np.ones(n), X))
```

```python
X
```

```python
w_hat = np.linalg.inv(X.T@X)@X.T@np.array(y)
```

```python
w_hat
```

```python
y_pred = X@w_hat
```

```python
y_pred
```

```python
y
```

```python
from sklearn.metrics import mean_squared_error

def rmse(y_test, y_pred):
    return np.sqrt(mean_squared_error(y_test, y_pred))    
```

```python
rmse(y, y_pred)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y, y_pred)
```

**С заменой обратной матрицы на псевдо-обратну.**

```python
w_hat_pinv = np.linalg.pinv(X.T@X)@X.T@np.array(y)
```

```python
y_pred_pinv = X@w_hat_pinv
```

```python
rmse(y, y_pred_pinv)
```

```python
r2_score(y, y_pred_pinv)
```

**С помощью готовой библиотеки.**

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(X, y)
y_pred1 = model.predict(X)
```

```python
rmse(y, y_pred1)
```

```python
r2_score(y, y_pred1)
```

**Вывод:**

Прогностическая модель во всех трех случаях дала одинаковые цифры результата и одинаково высокий результат прогнозирования.
